```python
import pyarrow as pa
import pyarrow.ipc as ipc

src_path = "/rds/general/project/lms-scott-raw/live/Ada/perturbation_modeling/T_perturb/tokenized_data/adipocytes_subset_obese_WL/dataset_all_src/obese.dataset/data-00000-of-00001.arrow"
tgt_path = "/rds/general/project/lms-scott-raw/live/Ada/perturbation_modeling/T_perturb/tokenized_data/adipocytes_subset_obese_WL/dataset_all_tgt/1_weightloss.dataset/data-00000-of-00001.arrow"

# Read src
with open(src_path, "rb") as f:
    reader = ipc.open_stream(f)
    src_table = reader.read_all()

all_src_ids = [
    val for sublist in src_table["input_ids"].to_pylist() for val in sublist
]
print(f"Src max token ID: {max(all_src_ids)}")
print(f"Src unique tokens: {len(set(all_src_ids))}")

# Read tgt
with open(tgt_path, "rb") as f:
    reader = ipc.open_stream(f)
    tgt_table = reader.read_all()

all_tgt_ids = [
    val for sublist in tgt_table["input_ids"].to_pylist() for val in sublist
]
print(f"Tgt max token ID: {max(all_tgt_ids)}")
print(f"Tgt unique tokens: {len(set(all_tgt_ids))}")
```
